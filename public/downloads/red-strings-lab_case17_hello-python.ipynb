## 2. Put the evidence in a list
A list holds many values in square brackets. Each number is one student's screen time in hours.

```python
screen_time = [4.0, 6.0, 2.5, 5.0, 3.5, 5.5, 3.0, 4.0, 5.0, 14.0]
print(screen_time)
print("How many answers:", len(screen_time))
```

## 3. The average
Add everything up, divide by how many. (Case 1 did this with a spreadsheet formula.)

```python
average = sum(screen_time) / len(screen_time)
print("Average:", average)
```

## 4. The median
The middle value once the list is sorted. Python's `statistics` module does it for us.

```python
import statistics
print("Sorted:", sorted(screen_time))
print("Median:", statistics.median(screen_time))
```

## 5. Find the weird number

```python
biggest = max(screen_time)
print("Biggest answer:", biggest)

without = [h for h in screen_time if h != biggest]
print("Average without it:", round(sum(without) / len(without), 2))
```
