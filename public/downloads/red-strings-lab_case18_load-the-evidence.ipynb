## 8. Does the link survive without the outlier?

```python
both = df["sleep"].corr(df["screen"])
without = df[df["screen"] < df["screen"].max()]
after = without["sleep"].corr(without["screen"])
print("Correlation, all ten:", round(both, 2))
print("Correlation, without the 14-hour answer:", round(after, 2))
```

Same as Case 1: about -0.37 with the outlier, about -0.96 without it. A link is a lead, not proof. **Say what you did with the outlier.**
